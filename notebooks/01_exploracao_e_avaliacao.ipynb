Avalia o classificador contra `categoria_real` (rótulos sintéticos) e inspeciona casos de baixa confiança.

```python
import sys; sys.path.insert(0, '..')
import pandas as pd
from src.classifier import TextClassifier
from src.config import Settings
from src.data_processing import classify_dataframe, generate_sample_data
```

```python
df = generate_sample_data(200)
clf = TextClassifier(Settings.from_env())
out = classify_dataframe(df, clf)
out.head()
```

```python
valid = out.dropna(subset=['categoria_prevista'])
print('Acurácia:', (valid.categoria_prevista == valid.categoria_real).mean())
pd.crosstab(valid.categoria_real, valid.categoria_prevista)
```
